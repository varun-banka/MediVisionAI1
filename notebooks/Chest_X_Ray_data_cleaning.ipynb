```python
import cv2

invalid = []

for folder in folders:

    split_path = os.path.join(DATASET_PATH, folder)

    for cls in os.listdir(split_path):

        class_path = os.path.join(split_path, cls)

        for file in os.listdir(class_path):

            path = os.path.join(class_path, file)

            img = cv2.imread(path)

            if img is None:
                continue

            h, w = img.shape[:2]

            if h == 0 or w == 0:
                invalid.append(path)

print("Invalid Images:", len(invalid))
```

```python
allowed = ["jpg", "jpeg", "png"]

invalid_format = []

for folder in folders:

    split_path = os.path.join(DATASET_PATH, folder)

    for cls in os.listdir(split_path):

        class_path = os.path.join(split_path, cls)

        for file in os.listdir(class_path):

            ext = file.split(".")[-1].lower()

            if ext not in allowed:
                invalid_format.append(file)

print("Invalid Formats:", len(invalid_format))
```

```python
def count_images(path):

    total = 0

    for cls in os.listdir(path):

        class_path = os.path.join(path, cls)

        total += len(os.listdir(class_path))

    return total

print("Train:", count_images(os.path.join(DATASET_PATH, "train")))
print("Validation:", count_images(os.path.join(DATASET_PATH, "val")))
print("Test:", count_images(os.path.join(DATASET_PATH, "test")))
```
